```python

# Keep only regular IPL innings (1 and 2)
ipl_regular = ipl_ball[
    ipl_ball["innings"].isin([1, 2])
].copy()

# Calculate total team runs for each delivery
ipl_regular["total_runs"] = (
    ipl_regular["runs"]
    + ipl_regular["extras1_runs"]
    + ipl_regular["extras_2_runs"]
)

# Identify deliveries that do not count as legal balls
ipl_regular["is_legal_ball"] = ~(
    ipl_regular["extras_type"].isin(["wide", "Overstep"])
)

print("Regular innings shape:", ipl_regular.shape)

print("\nExtras breakdown:")
extras_summary = (
    ipl_regular[ipl_regular["extras_type"].notna()]
    .groupby("extras_type")
    .agg(
        deliveries=("extras_type", "count"),
        batter_runs=("runs", "sum"),
        extra_runs_1=("extras1_runs", "sum"),
        extra_runs_2=("extras_2_runs", "sum"),
        total_runs=("total_runs", "sum")
    )
)
display(extras_summary)

print("\nLegal vs illegal deliveries:")
print(ipl_regular["is_legal_ball"].value_counts())

print("\nSample delivery calculations:")
display(
    ipl_regular[
        ipl_regular["extras_type"].notna()
    ][[
        "striker_name", "bowler_name", "runs",
        "extras_type", "extras1_runs",
        "extras_2_runs", "total_runs",
        "is_legal_ball"
    ]].head(15)
)
```

```python

# 1. Calculate the score and wickets in each regular innings
innings_summary = (
    ipl_regular.groupby(["match_id", "innings"])
    .agg(
        innings_runs=("total_runs", "sum"),
        wickets=("is_wicket", "sum"),
        legal_balls=("is_legal_ball", "sum")
    )
    .reset_index()
)

print("Sample innings summaries:")
display(innings_summary.head(10))


# 2. Get first-innings totals for each match
first_innings = innings_summary[
    innings_summary["innings"] == 1
][["match_id", "innings_runs"]].rename(
    columns={"innings_runs": "first_innings_total"}
)

print("\nFirst-innings total sample:")
display(first_innings.head(10))


# 3. Check whether any IPL matches have revised targets
revised_matches = ipl_matches[
    ipl_matches["revised_target"].notna()
][[
    "match_id", "comp_name", "season",
    "revised_target", "result"
]]

print("\nMatches with revised targets:", len(revised_matches))
display(revised_matches)


# 4. Inspect deliveries classified as 'other' extras
other_extras = ipl_regular[
    ipl_regular["extras_type"] == "other"
]

print("\nOther extras sample:")
display(
    other_extras[[
        "match_id", "innings", "striker_name",
        "bowler_name", "runs", "extras1_runs",
        "extras_2_runs", "delivery_style",
        "is_wicket"
    ]].head(20)
)
```

```python

# Test a stricter legal-ball definition:
# wides, Oversteps and "other" extras are excluded provisionally.

ipl_regular["legal_ball_v2"] = ~(
    ipl_regular["extras_type"].isin(["wide", "Overstep", "other"])
)

# Compare legal-ball counts using both definitions
ball_comparison = (
    ipl_regular.groupby(["match_id", "innings"])
    .agg(
        old_legal_balls=("is_legal_ball", "sum"),
        revised_legal_balls=("legal_ball_v2", "sum"),
        other_extras=("extras_type", lambda x: (x == "other").sum())
    )
    .reset_index()
)

print("Innings with more than 120 legal balls under the old rule:")
display(
    ball_comparison[ball_comparison["old_legal_balls"] > 120]
)

print("\nInnings with more than 120 legal balls under the revised rule:")
display(
    ball_comparison[ball_comparison["revised_legal_balls"] > 120]
)

# Examine all 'other' extras by delivery style
print("\nOther extras by delivery style:")
display(
    ipl_regular[ipl_regular["extras_type"] == "other"]
    .groupby("delivery_style", dropna=False)
    .agg(
        deliveries=("extras_type", "count"),
        extra_runs=("extras1_runs", "sum"),
        batter_runs=("runs", "sum")
    )
    .sort_values("deliveries", ascending=False)
)

# Inspect a few 'other' deliveries in detail
print("\nSample 'other' deliveries:")
display(
    ipl_regular[ipl_regular["extras_type"] == "other"][
        [
            "match_id", "innings", "overid",
            "over_ball_num", "innings_ball_num",
            "striker_name", "bowler_name",
            "runs", "extras1_runs", "extras_2_runs",
            "delivery_style"
        ]
    ].head(15)
)
```

```python

# Work on a copy of the regular IPL innings
pressure_data = ipl_regular.copy()

# Use our provisionally validated legal-ball rule
pressure_data["is_legal_ball"] = pressure_data["legal_ball_v2"]

# Ensure deliveries are in chronological innings order
pressure_data = pressure_data.sort_values(
    ["match_id", "innings", "innings_ball_num"]
).copy()

# Calculate match situation BEFORE each delivery
group_cols = ["match_id", "innings"]

pressure_data["score_before"] = (
    pressure_data.groupby(group_cols)["total_runs"].cumsum()
    - pressure_data["total_runs"]
)

pressure_data["wickets_before"] = (
    pressure_data.groupby(group_cols)["is_wicket"].cumsum()
    - pressure_data["is_wicket"].astype(int)
)

pressure_data["legal_balls_before"] = (
    pressure_data.groupby(group_cols)["is_legal_ball"].cumsum()
    - pressure_data["is_legal_ball"].astype(int)
)

# Standard T20 balls remaining
pressure_data["balls_remaining"] = (
    120 - pressure_data["legal_balls_before"]
)

# Add match information
match_info = ipl_matches[
    ["match_id", "comp_name", "season", "revised_target", "result"]
].drop_duplicates("match_id")

pressure_data = pressure_data.merge(
    match_info,
    on="match_id",
    how="left"
)

# Calculate first-innings score for each match
first_scores = (
    pressure_data[pressure_data["innings"] == 1]
    .groupby("match_id")["total_runs"]
    .sum()
    .reset_index(name="first_innings_total")
)

pressure_data = pressure_data.merge(
    first_scores,
    on="match_id",
    how="left"
)

# Set the second-innings target
# Use revised target when available; otherwise first innings total + 1
pressure_data["target"] = np.where(
    pressure_data["revised_target"].notna(),
    pressure_data["revised_target"],
    pressure_data["first_innings_total"] + 1
)

# Calculate chase situation
pressure_data["runs_required"] = (
    pressure_data["target"] - pressure_data["score_before"]
)

pressure_data["required_run_rate"] = np.where(
    (pressure_data["innings"] == 2)
    & (pressure_data["balls_remaining"] > 0)
    & (pressure_data["runs_required"] > 0),
    pressure_data["runs_required"] * 6
    / pressure_data["balls_remaining"],
    np.nan
)

# Display important columns
columns_to_check = [
    "match_id", "innings", "overid", "innings_ball_num",
    "striker_name", "bowler_name", "total_runs",
    "score_before", "wickets_before",
    "legal_balls_before", "balls_remaining",
    "target", "runs_required", "required_run_rate"
]

display(pressure_data[columns_to_check].head(20))
```

```python

# Inspect revised overs and targets for IPL matches
dls_check = ipl_matches[
    ipl_matches["revised_target"].notna()
][[
    "match_id",
    "season",
    "first_inn_revised_overs",
    "second_inn_revised_overs",
    "revised_target",
    "result"
]].copy()

print("DLS / revised-target matches:")
display(dls_check)

# Compare actual legal balls with the scheduled/revised overs
actual_balls = (
    innings_summary.pivot(
        index="match_id",
        columns="innings",
        values="legal_balls"
    )
    .rename(columns={
        1: "first_innings_legal_balls",
        2: "second_innings_legal_balls"
    })
    .reset_index()
)

overs_check = dls_check.merge(
    actual_balls,
    on="match_id",
    how="left"
)

print("\nActual legal balls in these matches:")
display(overs_check)
```

```python
# Rebuild innings summary using the revised legal-ball definition
innings_summary_v2 = (
    ipl_regular.groupby(["match_id", "innings"])
    .agg(
        innings_runs=("total_runs", "sum"),
        wickets=("is_wicket", "sum"),
        legal_balls=("legal_ball_v2", "sum")
    )
    .reset_index()
)

# Select matches with revised targets
dls_check = ipl_matches[
    ipl_matches["revised_target"].notna()
][[
    "match_id",
    "season",
    "first_inn_revised_overs",
    "second_inn_revised_overs",
    "revised_target"
]].copy()

# Get actual legal balls in each innings
actual_balls = (
    innings_summary_v2.pivot(
        index="match_id",
        columns="innings",
        values="legal_balls"
    )
    .rename(columns={
        1: "first_innings_legal_balls",
        2: "second_innings_legal_balls"
    })
    .reset_index()
)

# Compare revised limits with actual legal balls
overs_check_v2 = dls_check.merge(
    actual_balls,
    on="match_id",
    how="left"
)

display(overs_check_v2)
```

```python
# Get the maximum available balls for each innings
match_limits = ipl_matches[
    [
        "match_id",
        "first_inn_revised_overs",
        "second_inn_revised_overs"
    ]
].drop_duplicates("match_id").copy()

# Missing revised limits default to 120 balls
match_limits["first_innings_limit"] = (
    match_limits["first_inn_revised_overs"].fillna(120)
)

match_limits["second_innings_limit"] = (
    match_limits["second_inn_revised_overs"].fillna(120)
)

# Add innings limits to delivery-level pressure data
pressure_data = pressure_data.drop(
    columns=[
        "first_inn_revised_overs",
        "second_inn_revised_overs",
        "first_innings_limit",
        "second_innings_limit",
        "innings_ball_limit"
    ],
    errors="ignore"
)

pressure_data = pressure_data.merge(
    match_limits[
        [
            "match_id",
            "first_innings_limit",
            "second_innings_limit"
        ]
    ],
    on="match_id",
    how="left"
)

# Assign the correct limit based on the innings
pressure_data["innings_ball_limit"] = np.where(
    pressure_data["innings"] == 1,
    pressure_data["first_innings_limit"],
    pressure_data["second_innings_limit"]
)

# Calculate balls remaining
pressure_data["balls_remaining"] = (
    pressure_data["innings_ball_limit"]
    - pressure_data["legal_balls_before"]
)

# Required runs and required run rate are meaningful only in the chase
pressure_data["runs_required"] = np.where(
    pressure_data["innings"] == 2,
    pressure_data["target"] - pressure_data["score_before"],
    np.nan
)

pressure_data["required_run_rate"] = np.where(
    (pressure_data["innings"] == 2)
    & (pressure_data["balls_remaining"] > 0)
    & (pressure_data["runs_required"] > 0),
    pressure_data["runs_required"] * 6
    / pressure_data["balls_remaining"],
    np.nan
)

# Validate the calculations
print("Negative balls remaining:",
      (pressure_data["balls_remaining"] < 0).sum())

print("Missing innings limits:",
      pressure_data["innings_ball_limit"].isna().sum())

print("\nDLS match validation:")
display(
    pressure_data[
        pressure_data["match_id"].isin(
            [21586, 40727, 40749, 50137, 50371]
        )
    ].groupby(["match_id", "innings"])
    .agg(
        innings_limit=("innings_ball_limit", "first"),
        actual_legal_balls=("legal_balls_before", "max"),
        max_balls_remaining=("balls_remaining", "max"),
        min_balls_remaining=("balls_remaining", "min")
    )
    .reset_index()
)
```

```python
# Analyse pressure conditions during second-innings chases

chase_data = pressure_data[
    (pressure_data["innings"] == 2)
    & (pressure_data["balls_remaining"] > 0)
    & (pressure_data["runs_required"] > 0)
].copy()

print("Chase deliveries:", len(chase_data))

print("\nRequired Run Rate distribution:")
display(
    chase_data["required_run_rate"].describe(
        percentiles=[0.25, 0.50, 0.75, 0.90, 0.95]
    )
)

print("\nWickets lost distribution:")
display(
    chase_data["wickets_before"].value_counts()
    .sort_index()
)

print("\nRequired Run Rate by wickets lost:")
display(
    chase_data.groupby("wickets_before")["required_run_rate"]
    .agg(["count", "mean", "median", "max"])
    .round(2)
)
```

```python
# Inspect the 20 highest required run-rate situations
extreme_chases = chase_data.sort_values(
    "required_run_rate",
    ascending=False
).head(20)

display(
    extreme_chases[
        [
            "match_id",
            "season",
            "innings_ball_num",
            "score_before",
            "wickets_before",
            "balls_remaining",
            "runs_required",
            "required_run_rate",
            "striker_name",
            "bowler_name"
        ]
    ]
)
```

```python
# Create a provisional Chase Pressure Score

# Copy the chase data
chase_data = chase_data.copy()

# Cap extreme required run rates at the 95th percentile
rrr_cap = chase_data["required_run_rate"].quantile(0.95)

chase_data["rrr_capped"] = (
    chase_data["required_run_rate"].clip(upper=rrr_cap)
)

# Convert both factors into percentile ranks
chase_data["rrr_percentile"] = (
    chase_data["rrr_capped"].rank(pct=True)
)

chase_data["wickets_percentile"] = (
    chase_data["wickets_before"].rank(pct=True)
)

# Equal-weight pressure score (0-100)
chase_data["chase_pressure_score"] = (
    100 * (
        0.50 * chase_data["rrr_percentile"]
        + 0.50 * chase_data["wickets_percentile"]
    )
)

# Define pressure bands using score percentiles
p75 = chase_data["chase_pressure_score"].quantile(0.75)
p90 = chase_data["chase_pressure_score"].quantile(0.90)

chase_data["pressure_band"] = pd.cut(
    chase_data["chase_pressure_score"],
    bins=[-np.inf, 50, p75, p90, np.inf],
    labels=["Low", "Moderate", "High", "Extreme"],
    include_lowest=True
)

print("RRR cap (95th percentile):", round(rrr_cap, 2))
print("75th percentile pressure score:", round(p75, 2))
print("90th percentile pressure score:", round(p90, 2))

print("\nPressure band distribution:")
display(
    chase_data["pressure_band"]
    .value_counts()
    .sort_index()
)

print("\nSample high-pressure deliveries:")
display(
    chase_data.sort_values(
        "chase_pressure_score",
        ascending=False
    )[
        [
            "match_id",
            "striker_name",
            "bowler_name",
            "score_before",
            "wickets_before",
            "balls_remaining",
            "required_run_rate",
            "chase_pressure_score",
            "pressure_band"
        ]
    ].head(15)
)
```

```python
# Compare two approaches to handling extreme required run rates

rrr_cap_95 = chase_data["required_run_rate"].quantile(0.95)
rrr_cap_99 = chase_data["required_run_rate"].quantile(0.99)

# Calculate scores using the 99th percentile cap
chase_data["rrr_capped_99"] = (
    chase_data["required_run_rate"].clip(upper=rrr_cap_99)
)

chase_data["rrr_percentile_99"] = (
    chase_data["rrr_capped_99"].rank(pct=True)
)

chase_data["pressure_score_99"] = (
    100 * (
        0.50 * chase_data["rrr_percentile_99"]
        + 0.50 * chase_data["wickets_percentile"]
    )
)

# Compare score distributions
comparison = pd.DataFrame({
    "Method": ["95th percentile cap", "99th percentile cap"],
    "RRR cap": [rrr_cap_95, rrr_cap_99],
    "Unique pressure scores": [
        chase_data["chase_pressure_score"].nunique(),
        chase_data["pressure_score_99"].nunique()
    ],
    "Highest score": [
        chase_data["chase_pressure_score"].max(),
        chase_data["pressure_score_99"].max()
    ],
    "Deliveries with highest score": [
        (chase_data["chase_pressure_score"] ==
         chase_data["chase_pressure_score"].max()).sum(),
        (chase_data["pressure_score_99"] ==
         chase_data["pressure_score_99"].max()).sum()
    ]
})

display(comparison.round(2))

# Compare pressure bands using the 99th percentile version
p75_99 = chase_data["pressure_score_99"].quantile(0.75)
p90_99 = chase_data["pressure_score_99"].quantile(0.90)

chase_data["pressure_band_99"] = pd.cut(
    chase_data["pressure_score_99"],
    bins=[-np.inf, 50, p75_99, p90_99, np.inf],
    labels=["Low", "Moderate", "High", "Extreme"],
    include_lowest=True
)

print("\nPressure bands using 99th percentile cap:")
display(
    chase_data["pressure_band_99"]
    .value_counts()
    .sort_index()
)

print("\nTop 10 scores using 99th percentile cap:")
display(
    chase_data.nlargest(10, "pressure_score_99")[
        [
            "match_id",
            "striker_name",
            "wickets_before",
            "balls_remaining",
            "required_run_rate",
            "pressure_score_99"
        ]
    ]
)
```

```python
# Calculate pressure using the original, uncapped RRR

chase_data["rrr_percentile_raw"] = (
    chase_data["required_run_rate"].rank(pct=True)
)

chase_data["pressure_score_raw"] = (
    100 * (
        0.50 * chase_data["rrr_percentile_raw"]
        + 0.50 * chase_data["wickets_percentile"]
    )
)

# Compare all three approaches
comparison = pd.DataFrame({
    "Method": [
        "95th percentile cap",
        "99th percentile cap",
        "No cap - raw RRR percentile"
    ],
    "Unique scores": [
        chase_data["chase_pressure_score"].nunique(),
        chase_data["pressure_score_99"].nunique(),
        chase_data["pressure_score_raw"].nunique()
    ],
    "Maximum score": [
        chase_data["chase_pressure_score"].max(),
        chase_data["pressure_score_99"].max(),
        chase_data["pressure_score_raw"].max()
    ],
    "Deliveries tied at maximum": [
        (chase_data["chase_pressure_score"] ==
         chase_data["chase_pressure_score"].max()).sum(),
        (chase_data["pressure_score_99"] ==
         chase_data["pressure_score_99"].max()).sum(),
        (chase_data["pressure_score_raw"] ==
         chase_data["pressure_score_raw"].max()).sum()
    ]
})

display(comparison.round(2))

print("\nTop 10 situations using uncapped RRR percentile:")
display(
    chase_data.nlargest(10, "pressure_score_raw")[
        [
            "match_id",
            "striker_name",
            "wickets_before",
            "balls_remaining",
            "required_run_rate",
            "pressure_score_raw"
        ]
    ]
)
```

```python
# Finalize the uncapped pressure score
chase_data["pressure_score"] = (
    100 * (
        0.50 * chase_data["rrr_percentile_raw"]
        + 0.50 * chase_data["wickets_percentile"]
    )
)

# Define the top 10% as high-pressure situations
high_pressure_cutoff = chase_data["pressure_score"].quantile(0.90)

high_pressure = chase_data[
    chase_data["pressure_score"] >= high_pressure_cutoff
].copy()

# A wide is not counted as a ball faced by the batter
high_pressure["batter_faced_ball"] = (
    ~high_pressure["extras_type"].eq("wide")
)

# Keep only deliveries that count as balls faced
batter_exposure = high_pressure[
    high_pressure["batter_faced_ball"]
].copy()

# Calculate each batter's high-pressure exposure
batter_pressure_sample = (
    batter_exposure.groupby("striker_id")
    .agg(
        player_name=("striker_name", "first"),
        pressure_balls=("striker_id", "size"),
        matches=("match_id", "nunique")
    )
    .reset_index()
)

print("High-pressure cutoff:", round(high_pressure_cutoff, 2))
print("High-pressure deliveries:", len(high_pressure))
print("High-pressure balls faced by batters:", len(batter_exposure))
print("Unique batters:", batter_pressure_sample["striker_id"].nunique())

print("\nPlayer exposure distribution:")
display(
    batter_pressure_sample["pressure_balls"].describe(
        percentiles=[0.25, 0.50, 0.75, 0.90]
    )
)

print("\nPlayers meeting different sample-size thresholds:")
for threshold in [10, 20, 30, 50]:
    count = (
        batter_pressure_sample["pressure_balls"] >= threshold
    ).sum()
    print(f"{threshold}+ pressure balls: {count} players")

print("\nTop 20 batters by high-pressure exposure:")
display(
    batter_pressure_sample.sort_values(
        "pressure_balls",
        ascending=False
    ).head(20)
)
```

```python
# Correct the batting metrics using team runs for dot-ball detection

batter_exposure = batter_exposure.copy()

# A dot ball means no runs were scored by either the batter or as extras
batter_exposure["is_dot_ball"] = (
    batter_exposure["total_runs"] == 0
)

# Count only dismissals of the striker
batter_exposure["is_batter_dismissed"] = (
    batter_exposure["dismissed_player_id"]
    == batter_exposure["striker_id"]
)

# Recalculate batting statistics
batting_pressure_stats = (
    batter_exposure.groupby("striker_id")
    .agg(
        player_name=("striker_name", "first"),
        pressure_balls=("striker_id", "size"),
        matches=("match_id", "nunique"),
        runs=("runs", "sum"),
        fours=("is_four", "sum"),
        sixes=("is_six", "sum"),
        dot_balls=("is_dot_ball", "sum"),
        dismissals=("is_batter_dismissed", "sum")
    )
    .reset_index()
)

# Calculate performance metrics
batting_pressure_stats["strike_rate"] = (
    batting_pressure_stats["runs"]
    / batting_pressure_stats["pressure_balls"] * 100
)

batting_pressure_stats["boundary_rate"] = (
    (batting_pressure_stats["fours"] +
     batting_pressure_stats["sixes"])
    / batting_pressure_stats["pressure_balls"] * 100
)

batting_pressure_stats["dot_ball_rate"] = (
    batting_pressure_stats["dot_balls"]
    / batting_pressure_stats["pressure_balls"] * 100
)

batting_pressure_stats["dismissal_rate"] = (
    batting_pressure_stats["dismissals"]
    / batting_pressure_stats["pressure_balls"] * 100
)

# Apply the provisional eligibility rule
eligible_batters = batting_pressure_stats[
    (batting_pressure_stats["pressure_balls"] >= 30)
    & (batting_pressure_stats["matches"] >= 3)
].copy()

print("Eligible batters:", len(eligible_batters))

print("\nCorrected batting performance:")
display(
    eligible_batters.sort_values(
        "pressure_balls",
        ascending=False
    )[
        [
            "player_name",
            "pressure_balls",
            "matches",
            "runs",
            "strike_rate",
            "boundary_rate",
            "dot_ball_rate",
            "dismissals",
            "dismissal_rate"
        ]
    ].round(2)
)

print("\nTop 15 by corrected pressure strike rate:")
display(
    eligible_batters.sort_values(
        "strike_rate",
        ascending=False
    )[
        [
            "player_name",
            "pressure_balls",
            "matches",
            "runs",
            "strike_rate",
            "dot_ball_rate",
            "dismissals"
        ]
    ].head(15).round(2)
)
```

```python
# Calculate each player's performance in each high-pressure match
match_batting = (
    batter_exposure.groupby(["striker_id", "striker_name", "match_id"])
    .agg(
        pressure_balls=("striker_id", "size"),
        runs=("runs", "sum"),
        dismissals=("is_batter_dismissed", "sum")
    )
    .reset_index()
)

match_batting["strike_rate"] = (
    match_batting["runs"] / match_batting["pressure_balls"] * 100
)

# Summarise match-level consistency for eligible batters
consistency = (
    match_batting[
        match_batting["striker_id"].isin(
            eligible_batters["striker_id"]
        )
    ]
    .groupby("striker_id")
    .agg(
        matches_with_pressure=("match_id", "nunique"),
        matches_with_5plus_balls=(
            "pressure_balls",
            lambda x: (x >= 5).sum()
        ),
        matches_with_5plus_runs=(
            "runs",
            lambda x: (x >= 5).sum()
        ),
        average_runs_per_match=("runs", "mean"),
        median_runs_per_match=("runs", "median")
    )
    .reset_index()
)

consistency = consistency.merge(
    eligible_batters[["striker_id", "player_name"]],
    on="striker_id",
    how="left"
)

consistency["contribution_match_rate"] = (
    consistency["matches_with_5plus_runs"]
    / consistency["matches_with_pressure"] * 100
)

display(
    consistency.sort_values(
        ["contribution_match_rate", "matches_with_pressure"],
        ascending=False
    )[
        [
            "player_name",
            "matches_with_pressure",
            "matches_with_5plus_balls",
            "matches_with_5plus_runs",
            "average_runs_per_match",
            "contribution_match_rate"
        ]
    ].round(2)
)
```

```python
# Create percentile-based component scores

score_data = eligible_batters.copy()

# Higher is better
score_data["sr_score"] = (
    score_data["strike_rate"].rank(pct=True) * 100
)

score_data["boundary_score"] = (
    score_data["boundary_rate"].rank(pct=True) * 100
)

# Lower is better, so reverse the percentile
score_data["dot_score"] = (
    (1 - score_data["dot_ball_rate"].rank(pct=True)) * 100
)

score_data["survival_score"] = (
    (1 - score_data["dismissal_rate"].rank(pct=True)) * 100
)

# Add match consistency
score_data = score_data.merge(
    consistency[
        ["striker_id", "contribution_match_rate"]
    ],
    on="striker_id",
    how="left"
)

score_data["consistency_score"] = (
    score_data["contribution_match_rate"]
)

# Weighted reliability score
score_data["reliability_score"] = (
    0.30 * score_data["sr_score"]
    + 0.15 * score_data["boundary_score"]
    + 0.15 * score_data["dot_score"]
    + 0.20 * score_data["survival_score"]
    + 0.20 * score_data["consistency_score"]
)

# Display top 15
top_batters = score_data.sort_values(
    "reliability_score",
    ascending=False
)

display(
    top_batters[
        [
            "player_name",
            "pressure_balls",
            "matches",
            "runs",
            "strike_rate",
            "dot_ball_rate",
            "dismissal_rate",
            "contribution_match_rate",
            "reliability_score"
        ]
    ].head(15).round(2)
)
```

```python
# 1. Calculate match-level bowling performance
match_bowling = (
    bowling_data.groupby(
        ["bowler_id", "bowler_name", "match_id"]
    )
    .agg(
        legal_balls=("bowler_legal_ball", "sum"),
        runs_conceded=("bowler_runs_conceded", "sum"),
        wickets=("bowler_wicket", "sum")
    )
    .reset_index()
)

# Consider matches with at least 6 legal pressure balls
qualified_bowling_matches = match_bowling[
    match_bowling["legal_balls"] >= 6
].copy()

qualified_bowling_matches["match_economy"] = (
    qualified_bowling_matches["runs_conceded"] * 6
    / qualified_bowling_matches["legal_balls"]
)

# Overall median match-level economy benchmark
economy_benchmark = qualified_bowling_matches[
    "match_economy"
].median()

qualified_bowling_matches["controlled_match"] = (
    qualified_bowling_matches["match_economy"]
    <= economy_benchmark
)

# 2. Calculate consistency for each bowler
bowling_consistency = (
    qualified_bowling_matches.groupby("bowler_id")
    .agg(
        qualified_matches=("match_id", "nunique"),
        controlled_matches=("controlled_match", "sum")
    )
    .reset_index()
)

bowling_consistency["consistency_rate"] = (
    bowling_consistency["controlled_matches"]
    / bowling_consistency["qualified_matches"] * 100
)

# 3. Add consistency to eligible bowlers
bowler_score_data = eligible_bowlers.merge(
    bowling_consistency,
    on="bowler_id",
    how="left"
)

bowler_score_data["consistency_rate"] = (
    bowler_score_data["consistency_rate"].fillna(0)
)

# 4. Convert metrics into percentile scores
bowler_score_data["economy_score"] = (
    1 - bowler_score_data["economy"].rank(pct=True)
) * 100

bowler_score_data["wicket_score"] = (
    bowler_score_data["wickets_per_100_balls"].rank(pct=True)
) * 100

bowler_score_data["dot_score"] = (
    bowler_score_data["dot_ball_rate"].rank(pct=True)
) * 100

bowler_score_data["boundary_score"] = (
    1 - bowler_score_data["boundary_rate"].rank(pct=True)
) * 100

bowler_score_data["consistency_score"] = (
    bowler_score_data["consistency_rate"].rank(pct=True)
) * 100

# 5. Calculate weighted reliability score
bowler_score_data["reliability_score"] = (
    0.25 * bowler_score_data["economy_score"]
    + 0.25 * bowler_score_data["wicket_score"]
    + 0.20 * bowler_score_data["dot_score"]
    + 0.15 * bowler_score_data["boundary_score"]
    + 0.15 * bowler_score_data["consistency_score"]
)

# 6. Display the preliminary bowling rankings
bowler_rankings = bowler_score_data.sort_values(
    "reliability_score",
    ascending=False
).reset_index(drop=True)

bowler_rankings.insert(
    0, "rank", range(1, len(bowler_rankings) + 1)
)

print("Match economy benchmark:", round(economy_benchmark, 2))
print("Eligible bowlers scored:", len(bowler_rankings))

display(
    bowler_rankings[
        [
            "rank", "player_name", "legal_balls", "matches",
            "wickets", "economy", "wickets_per_100_balls",
            "dot_ball_rate", "boundary_rate",
            "consistency_rate", "reliability_score"
        ]
    ].head(20).round(2)
)
```

```python

# Prepare batting data with revised consistency
batting_sample_data = batting_pressure_stats.copy()

batting_sample_data = batting_sample_data.merge(
    batting_consistency_v2[
        [
            "striker_id",
            "qualified_matches",
            "qualified_contribution_rate"
        ]
    ],
    on="striker_id",
    how="left"
)

batting_sample_data["qualified_matches"] = (
    batting_sample_data["qualified_matches"].fillna(0)
)

# Define sample-size scenarios
sample_scenarios = {
    "30 balls / 3 matches": (30, 3),
    "40 balls / 4 matches": (40, 4),
    "50 balls / 5 matches": (50, 5)
}

batting_threshold_results = {}

for scenario, (min_balls, min_matches) in sample_scenarios.items():

    temp = batting_sample_data[
        (batting_sample_data["pressure_balls"] >= min_balls)
        & (batting_sample_data["matches"] >= min_matches)
        & (batting_sample_data["qualified_matches"] >= 3)
    ].copy()

    # Recalculate percentile scores within each eligible group
    temp["sr_score"] = temp["strike_rate"].rank(pct=True) * 100

    temp["boundary_score"] = (
        temp["boundary_rate"].rank(pct=True) * 100
    )

    temp["dot_score"] = (
        (1 - temp["dot_ball_rate"].rank(pct=True)) * 100
    )

    temp["survival_score"] = (
        (1 - temp["dismissal_rate"].rank(pct=True)) * 100
    )

    temp["consistency_score"] = (
        temp["qualified_contribution_rate"]
    )

    # Balanced reliability score
    temp["reliability_score"] = (
        0.30 * temp["sr_score"]
        + 0.15 * temp["boundary_score"]
        + 0.15 * temp["dot_score"]
        + 0.20 * temp["survival_score"]
        + 0.20 * temp["consistency_score"]
    )

    temp = temp.sort_values(
        "reliability_score",
        ascending=False
    ).reset_index(drop=True)

    temp.insert(0, "rank", range(1, len(temp) + 1))

    batting_threshold_results[scenario] = temp

    print(f"\n{scenario} — Eligible players: {len(temp)}")

    display(
        temp[
            [
                "rank",
                "player_name",
                "pressure_balls",
                "matches",
                "qualified_matches",
                "strike_rate",
                "reliability_score"
            ]
        ].head(10).round(2)
    )
```
